```python
import pandas as pd
```

```python
df = pd.read_csv('friends_quotes.csv')
```

```python
# CRITERIOS DE BUSQUEDA PARA BODAS
bodas = ["Wedding",               # Atrapa casi todas las bodas
    "Monica Gets A Roommate", # Boda fallida de Rachel
    "In Vegas",    # Boda Ross y Rachel
    "The Last One"]           # Escenas finales

# CONVERTIR LA LISTA EN FILTRO
# El método .join une la lista con el símbolo '|' que en programación significa "OR" (O).
# Esto crea el texto: "Wedding|Vegas|Monica Gets A Roommate|The Last One"
patron_busqueda = '|'.join(bodas)

# FILTRAR USANDO LA LISTA
df_bodas = df[df['episode_title'].str.contains(patron_busqueda, case=False, na=False)]

# GUARDAR NUEVO CSV
df_bodas.to_csv('friends_quotes_clean.csv', index=False)

print(f"Filas finales: {len(df_bodas)}")
```
